## Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Membaca Dataset

```python
df = pd.read_csv("indonesia_mbg_nationwide_enriched_2026.csv")

df.head()
```

## Melihat Informasi dan Deskripsi Data

```python
df.info()
df.describe()
```

## Menghitung data kosong pada kolom

```python
df.isna().sum()
```

## Menampilkan unik dari kolom kategorik

```python
df["province"].unique()
```

## Isi data kosong dengan angka

```python
for col in df.select_dtypes(include=["float64", "int64"]).columns:
  df[col].fillna(df[col].mean())

for col in df.select_dtypes(include=["object"]).columns:
  if df[col].isna().sum() > 0:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("Data kosong setelah diisi:")
print(df.isna().sum().sum())
```

## Hitung data duplikat

```python
print("Jumlah data duplikat:", df.duplicated().sum())
```

## Hapus data duplikat dan cek kembali

```python
df.drop_duplicates()
df.info()
```

## Visualisasi dengan grafik batang

```python
plt.figure(figsize=(10, 6))

df["bgn_program_compliance_tier"].value_counts().plot(
    kind="bar", color="skyblue", edgecolor="black"
)

plt.title(
    "Distribusi Kategori Kepatuhan Program BGN (Compliance Tier)",
    fontsize=14,
    pad=15,
)
plt.xlabel("Kategori Kepatuhan", fontsize=12)
plt.ylabel("Jumlah Frekuensi", fontsize=12)

plt.xticks(rotation=45, ha="right")

plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
