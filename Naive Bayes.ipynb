```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('StudyCase - Anggota_Perpus.csv')
print("--- 5 Data Teratas ---")
display(df.head())
```

```python
print("\n--- Informasi Dataset (info()) ---")
df.info()

print("\n--- Deskripsi Statistik (describe()) ---")
display(df.describe(include='all'))
```

```python
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"\nJumlah Fitur Numerik: {len(numeric_features)}")
print(f"Daftar Fitur Numerik: {numeric_features}")
```

```python
categorical_features = df.select_dtypes(include=['object', 'category']).columns.tolist()
print(f"\nJumlah Fitur Kategorikal: {len(categorical_features)}")
print(f"Daftar Fitur Kategorikal: {categorical_features}")
```

```python
print("\n--- Nilai Unik Fitur Kategorikal ---")
for col in categorical_features:
    print(f"Kolom '{col}' memiliki {df[col].nunique()} nilai unik.")
```

```python
df_encoded = df.copy()
label_encoders = {}

for col in categorical_features:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))
    label_encoders[col] = le

print("\n--- Hasil Dataset Setelah LabelEncoder (head()) ---")
display(df_encoded.head())
```
